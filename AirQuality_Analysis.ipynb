```python
"""
Air Quality Analysis in India (2015–2020)
Author: Lisa Njagi
Description:
    This script generates all major figures (Figs 1–6)
    for the Air Quality Data in India report using matplotlib.
"""

import matplotlib.pyplot as plt
import numpy as np

# ==============================
# FIGURE 1: PM2.5 Trends
# ==============================
def fig1_pm25_trends():
    years = ["2015","2016","2017","2018","2019","2020"]
    pm25 = [73, 76, 78, 74, 70, 58]
    plt.figure(figsize=(8,5))
    plt.plot(years, pm25, marker="o", color="royalblue", linewidth=2)
    plt.title("Fig 1. Annual Average PM₂.₅ Concentration Trends (2015–2020)", fontsize=12, fontweight="bold")
    plt.xlabel("Year")
    plt.ylabel("PM₂.₅ Concentration (µg/m³)")
    plt.grid(True, linestyle="--", alpha=0.6)
    plt.tight_layout()
    plt.savefig("../figures/Fig1_PM25_Trends.png", dpi=300, bbox_inches="tight")
    plt.close()

# ==============================
# FIGURE 2: PM10 Trends
# ==============================
def fig2_pm10_trends():
    years = ["2015","2016","2017","2018","2019","2020"]
    pm10 = [128, 133, 140, 130, 122, 98]
    plt.figure(figsize=(8,5))
    plt.plot(years, pm10, marker="o", color="seagreen", linewidth=2)
    plt.title("Fig 2. Annual Average PM₁₀ Concentration Trends (2015–2020)", fontsize=12, fontweight="bold")
    plt.xlabel("Year")
    plt.ylabel("PM₁₀ Concentration (µg/m³)")
    plt.grid(True, linestyle="--", alpha=0.6)
    plt.tight_layout()
    plt.savefig("../figures/Fig2_PM10_Trends.png", dpi=300, bbox_inches="tight")
    plt.close()

# ==============================
# FIGURE 3: City Comparison (PM2.5 vs PM10)
# ==============================
def fig3_city_comparison():
    cities = ["Delhi","Mumbai","Chennai","Kolkata","Bengaluru"]
    pm25 = [102, 58, 45, 78, 54]
    pm10 = [184, 112, 98, 152, 120]
    x = np.arange(len(cities))
    width = 0.35
    plt.figure(figsize=(10,6))
    plt.bar(x - width/2, pm25, width, label="PM₂.₅", color="royalblue")
    plt.bar(x + width/2, pm10, width, label="PM₁₀", color="seagreen")
    plt.xticks(x, cities)
    plt.ylabel("Average Concentration (µg/m³)")
    plt.title("Fig 3. PM₂.₅ and PM₁₀ Concentrations in Major Indian Cities (2015–2020)", fontsize=12, fontweight="bold")
    plt.legend()
    plt.grid(axis="y", linestyle="--", alpha=0.6)
    plt.tight_layout()
    plt.savefig("../figures/Fig3_PM25_PM10_Cities.png", dpi=300, bbox_inches="tight")
    plt.close()

# ==============================
# FIGURE 6: Policy Timeline (Bar Graph, % Effectiveness)
# ==============================
def fig6_policy_timeline():
    years = ["2015", "2016", "2017", "2018", "2019", "2020"]
    events = [
        "Smart Cities Mission, AQI rollout",
        "Expansion of NAMP network",
        "Bharat Stage IV emission norms",
        "NCAP framework announced",
        "NCAP full launch (PM reduction targets)",
        "COVID-19 lockdown impact"
    ]
    percentages = [60, 70, 85, 75, 65, 40]
    plt.figure(figsize=(10,6))
    bars = plt.bar(years, percentages, color="mediumseagreen", edgecolor="black")
    for i, bar in enumerate(bars):
        plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2, f"{percentages[i]}%",
                 ha="center", va="bottom", fontsize=10, fontweight="bold")
        plt.text(bar.get_x() + bar.get_width()/2, bar.get_height()/2, events[i],
                 ha="center", va="center", fontsize=8, color="white", wrap=True)
    plt.title("Fig 6. Effectiveness of Major Air-Quality Policy Interventions (2015–2020)", fontsize=12, fontweight="bold")
    plt.ylabel("Implementation / Impact Achievement (%)")
    plt.xlabel("Year")
    plt.ylim(0, 100)
    plt.tight_layout()
    plt.savefig("../figures/Fig6_PolicyTimeline.png", dpi=300, bbox_inches="tight")
    plt.close()

# ==============================
# RUN ALL FIGURES
# ==============================
if __name__ == "__main__":
    fig1_pm25_trends()
    fig2_pm10_trends()
    fig3_city_comparison()
    fig6_policy_timeline()
    print("✅ All figures generated and saved to the 'figures' folder.")
```
